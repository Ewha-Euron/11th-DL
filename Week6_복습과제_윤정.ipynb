# **U-Net 모델 구현**

✅ 아래 코드는 PyTorch로 작성한 간단한 U-Net 모델입니다.

✅ 주요 목표: 작은 이미지 세트로 segmentation 학습을 테스트하기!

✅ 특징: 업샘플링(deconv/upsample 선택 가능), 입력 스케일링, Gaussian noise 추가, 배치 정규화 적용

```python

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset

# --- 작은 U-Net 정의 ---
class UNetSmall(nn.Module):
    def __init__(self, upsample_mode='deconv', net_scaling=None, edge_crop=0, gaussian_noise_std=0.1):
        super().__init__()
        self.net_scaling = net_scaling
        self.edge_crop = edge_crop
        self.gaussian_noise_std = gaussian_noise_std
        self.upsample_mode = upsample_mode

        # Convolution Block: Conv -> ReLU -> Conv -> ReLU
        def conv_block(in_c, out_c):
            return nn.Sequential(
                nn.Conv2d(in_c, out_c, 3, padding=1),   # 입력 채널 in_c → 출력 채널 out_c, 3x3 커널 (padding=1이라 크기 유지)
                nn.ReLU(inplace=True),
                nn.Conv2d(out_c, out_c, 3, padding=1),  # 두 번째 conv는 out_c → out_c (채널 수 유지)
                nn.ReLU(inplace=True)
            )

        # 인코더 부분 목표 -> 128 까지!
        self.enc1 = conv_block(3, 8)      # RGB 3채널 → 8
        self.enc2 = conv_block(8, 16)     # 내려갈수록 채널 2배씩 증가
        self.enc3 = conv_block(16, 32)
        self.enc4 = conv_block(32, 64)
        # Bottleneck
        self.center = conv_block(64, 128) # 가장 깊은 곳에서 128채널 도달
        self.pool = nn.MaxPool2d(2)       # 2x2 풀링 → 가로·세로 크기 절반


        # 업샘플링 부분 정의 (deconv 또는 upsample+conv) 로
        if upsample_mode == 'deconv':
            def upsample_layer(in_c, out_c):
                return nn.ConvTranspose2d(in_c, out_c, 2, stride=2)  # 커널 2, stride 2 → 크기 정확히 2배 (stride 없으면 크기 안 맞아 오류)
        else:
            def upsample_layer(in_c, out_c):
                return nn.Sequential(
                    nn.Upsample(scale_factor=2, mode='nearest'),  # 크기 2배로 단순 확대
                    nn.Conv2d(in_c, out_c, 1)                     # 1x1 conv로 채널 수만 조절
                )
        # 디코더 부분
        self.up4 = upsample_layer(128, 64)  # 128 → 64채널, 크기 2배
        self.dec4 = conv_block(128, 64)     # concat(up4 64 + e4 64) = 128 → 64
        self.up3 = upsample_layer(64, 32)
        self.dec3 = conv_block(64, 32)      # concat(32 + e3 32) = 64 → 32
        self.up2 = upsample_layer(32, 16)
        self.dec2 = conv_block(32, 16)      # concat(16 + e2 16) = 32 → 16
        self.up1 = upsample_layer(16, 8)
        self.dec1 = conv_block(16, 8)       # concat(8 + e1 8) = 16 → 8

        # 최종 출력: 1채널(segmentation mask)
        self.final = nn.Conv2d(8, 1, 1)     # 1x1 conv로 8채널 → 1채널 마스크

    def forward(self, x):
        if self.net_scaling:
        # 입력 스케일링
            x = F.avg_pool2d(x, self.net_scaling)
        # Gaussian noise 추가
        if self.gaussian_noise_std:
            noise = torch.randn_like(x) * self.gaussian_noise_std
            x = x + noise

        # 배치 정규화 적용
        x = F.batch_norm(x, running_mean=None, running_var=None, training=True)

        # 인코더
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(self.pool(e2))
        e4 = self.enc4(self.pool(e3))
        # Bottleneck
        center = self.center(self.pool(e4))   # e4를 한 번 더 풀링해서 bottleneck에 넣음 (16→8)

        # 디코더 + Skip Connection
        d4 = self.up4(center)
        d4 = torch.cat([d4, e4], dim=1)       # 같은 크기의 인코더 출력 e4를 채널 방향으로 이어붙임
        d4 = self.dec4(d4)

        d3 = self.up3(d4)
        d3 = torch.cat([d3, e3], dim=1)       # skip connection: e3
        d3 = self.dec3(d3)

        d2 = self.up2(d3)
        d2 = torch.cat([d2, e2], dim=1)       # skip connection: e2
        d2 = self.dec2(d2)

        d1 = self.up1(d2)
        d1 = torch.cat([d1, e1], dim=1)       # skip connection: e1
        d1 = self.dec1(d1)

        out = self.final(d1)
        out = torch.sigmoid(out)
         # Crop + Pad
        if self.edge_crop > 0:
            out = out[:, :, self.edge_crop:-self.edge_crop, self.edge_crop:-self.edge_crop]
            out = F.pad(out, (self.edge_crop, self.edge_crop, self.edge_crop, self.edge_crop))
        # Upsample 다시 확대
        if self.net_scaling:
            out = F.interpolate(out, scale_factor=self.net_scaling, mode='nearest')

        return out
```

## **학습용 데이터 생성하기**

```python
batch_size = 32
x = torch.randn(batch_size, 3, 128, 128)  # RGB 이미지
y = (torch.rand(batch_size, 1, 128, 128) > 0.5).float()  # 랜덤 이진 마스크

dataset = TensorDataset(x, y)
loader = DataLoader(dataset, batch_size=2, shuffle=True)
```

## **데이터 시각화**

```python
import matplotlib.pyplot as plt
# 첫 번째 이미지를 시각화
sample_x = x[0]
sample_y = y[0]

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.imshow(sample_x.permute(1, 2, 0).numpy())
plt.title('Input Image')

plt.subplot(1, 2, 2)
plt.imshow(sample_y.squeeze().numpy(), cmap='gray')
plt.title('Ground Truth Mask')
plt.show()
```

## **모델 선언 및 학습**

```python
# --- 모델 선언 및 학습 ---
model = UNetSmall(upsample_mode='deconv', net_scaling=None, edge_crop=0)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
criterion = nn.BCELoss()

# 간단히 30 epoch만
for epoch in range(10):
    model.train()
    total_loss = 0
    for xb, yb in loader:
        pred = model(xb)
        loss = criterion(pred, yb)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    print(f"Epoch {epoch+1}, Loss: {total_loss/len(loader):.4f}")
```

## **아래 주석을 풀고 U-Net 모델에 대해 간단히 설명해주세요 😀**

[답변]

**U-Net**은 의료영상 분할을 위해 제안된 **인코더–디코더 구조**의 segmentation 모델입니다 (모양이 U자).

- **인코더(수축 경로)**: Conv 블록 + MaxPooling을 반복해 해상도는 줄이고 채널은 늘리며 **"무엇이 있는지"(의미 정보)** 를 추출합니다. (이 코드: 8→16→32→64, bottleneck 128)
- **디코더(확장 경로)**: 업샘플링(ConvTranspose 또는 Upsample+Conv)으로 해상도를 다시 키워 **픽셀 단위 마스크**를 복원합니다.
- **Skip Connection**: 인코더의 같은 해상도 feature map을 디코더에 concat하여, 다운샘플링 과정에서 사라진 **"어디에 있는지"(위치·경계 정보)** 를 보충합니다. → 경계가 정밀해지고 적은 데이터로도 학습이 잘 됩니다.
- **출력**: 1×1 Conv로 1채널을 만든 뒤 sigmoid → 픽셀별 전경 확률, BCE loss로 학습합니다.



## **아래 주석을 풀고 DeepLabv1, v2, v3. v3+  모델에 대해 각각 간단히 설명해주세요 🐶**

[답변]

DeepLab 시리즈는 **Atrous(Dilated) Convolution**으로 해상도를 덜 잃으면서 넓은 receptive field를 얻는 semantic segmentation 모델입니다.

---

[DeepLab v1]
- VGG-16 backbone에 **Atrous Convolution**을 도입해 pooling으로 인한 해상도 손실을 줄이면서 receptive field를 확장.
- 출력의 거친 경계를 **Fully-Connected CRF** 후처리로 정교하게 다듬음.

[DeepLab v2]
- **ASPP(Atrous Spatial Pyramid Pooling)** 도입: dilation rate가 다른 atrous conv를 병렬로 적용해 **다양한 크기의 객체(멀티스케일)** 를 포착.
- backbone을 **ResNet-101**로 교체, CRF 후처리는 유지.

[DeepLab v3]
- ASPP 개선: **1×1 conv + Image-level pooling(전역 정보)** 추가, **Batch Normalization** 적용.
- 성능이 충분히 좋아져 **CRF 후처리 제거**. Cascaded atrous 모듈(multi-grid)도 사용.

[DeepLab v3+]
- v3를 인코더로 쓰고 **간단한 디코더**를 추가한 **Encoder–Decoder 구조** → low-level feature와 결합해 **객체 경계를 더 정확히** 복원.
- **Xception backbone + Depthwise Separable (Atrous) Convolution**으로 연산량을 줄이면서 성능 향상.
